#### Functions For preprocessing various columns

##### Clean Rating Column

````python
# We can see that unique values of rating column have float values and a few Nan values. The Nan values are to be changed into the mean rating of respective column

def clean_rating_column():
    """
    Cleans the 'Rating' column:
    - Replaces NaN with mean of respective category
    """
    try:
        #Using vectorized operation to fill Nan values in 'Rating' with mean value of all ratings of the respective 'Category'
        apps['Rating'] = apps['Rating'].fillna(apps.groupby('Category')['Rating'].transform('mean'))
    except Exception as e:
        print(f'Error Occured: {e}')
````

##### clean_reviews_column

````python
# Function to clean the 'Reviews' column

def clean_reviews_column():
    """
    Cleans the 'Reviews' column:
    - Converts all values to integers.
    """
    try:
        apps['Reviews'] = apps['Reviews'].astype(int)
    except Exception as e:
        print(f'Error Occured: {e}')
````

##### convert_size_column

````python
# Return rows in apps where the 'Size' column has values other than ones ending with 'M' or 'k"

print("Rows in 'apps' where 'Size' column has values other than ones ending with 'M' or 'k':")
display(apps[~apps['Size'].str.endswith(('M', 'k'))].shape[0])
````

````python
apps[apps['Size'] == 'Varies with device'].shape[0]
````

All the rows in apps where the 'Size' column has values other than ones ending with 'M' or 'k - they have value 'Varies with Device"

````python
# This function fills the null values in Size_MB with median values of each group
def fill_size_column():
    try:
        apps['Size_MB'] = apps['Size_MB'].fillna(apps.groupby('Category')['Size_MB'].transform('median'))
    except Exception as e:
        print(f'Error Occured: {e}')
````

````python
# Converting column values to float for M and K and values
# Converting size = 'Varies with device' to np.nan for further processing, then we update it with mean value of the sizes of apps in that 'Category'
def convert_size_column():

    cleaned_size = []

    for index, review in apps['Size'].items():
        try:
            review_str = review.strip()
            if review_str.endswith('M'):
                num = review_str[:-1]
                cleaned_size.append(round(float(num),2))
            elif review_str.endswith('k'):
                num = review_str[:-1]
                cleaned_size.append(round(float(num)/1024,2))
            else:
                cleaned_size.append(np.nan)
        except Exception as e:
            print(f'Error Occured: {e}')

    apps['Size_MB'] = cleaned_size

    # This function fills the null values in Size_MB with median values of each group
    fill_size_column()
````

##### clean_installs_column

````python
def clean_installs_column():
    """
    Cleans the 'Installs' column by removing ',' and '+' (only if present),
    then validates numeric format. Returns updated DataFrame and list of row indices that raised errors.
    """
    cleaned_installs = []

    for i, val in apps['Installs'].items():
        try:
            # Step 1: Convert to string safely
            val_str = str(val).replace(',', '').replace('+', '').strip()

            # Step 2: Validate that the cleaned string is all digits
            if not val_str.isdigit():
                raise ValueError(f"Invalid numeric format after cleaning: '{val_str}'")

            # Step 3: Convert to integer
            cleaned_installs.append(int(val_str))

        except Exception as e:
            print(f'Error Occured: {e}')

    apps['Installs'] = cleaned_installs
````

##### clean_price_column

````python
# View number of rows where the 'Price' column value does not start with '$'
rows_with_unusual_price = apps[~apps['Price'].str.startswith('$')].shape[0]

# view number of rows where 'Price' is 0
rows_with_0_price = apps[apps['Price'].str.isdigit()].shape[0]

print(rows_with_unusual_price)
print(rows_with_0_price)
````

We can see that the price is either 0 or in dollars.

````python
# If value in 'Price' starts with a $ then remove the $, otherwise change it to 0.0

def clean_price_column():
    """
    Cleans the 'Price' column by removing '$' and converting to float.
    Sets price to 0.0 if it doesn't start with '$'.
    """
    try:
        apps['Price'] = apps['Price'].apply(lambda x: x[1:] if x.startswith('$') else '0.0')
        apps['Price'] = apps['Price'].astype(float)
    except Exception as e:
        print(f'Error Occured: {e}')
````

##### clean_genres_column

````python
# View all rows in apps where the 'Genres' column shows two or more genres separated by a ';'

apps[apps['Genres'].str.contains(';', na=False)]
````

````python
# Function to split the genres (separated by ';') in apps into a list of genres. Even if there is only 1 genre, we still convert it into a list.

def split_genres():
    """
    Splits a string of genres separated by ';' into a list of genres.
    If there's only one genre, it's still returned as a list with one element.
    Handles potential NaN values by returning an empty list.
    """
    apps['Genres List'] = apps['Genres'].apply(lambda x: x.split(';'))
````

##### convert_last_updated

````python
# Check for any null values in 'Last Updated' in apps

print(apps['Last Updated'].isnull().sum())
````

````python
# Extract the 'year' value into a new column - 'Last Updated Year' for visualization.

def convert_last_updated():
    """
    Converts the 'Last Updated' column to datetime objects.
    Handles potential errors during conversion.
    Extracts the year from the date and stores it in a new column 'Last Updated Year'.
    """
    try:
        apps['Last Updated Year'] = pd.to_datetime(apps['Last Updated'], errors='coerce').dt.year
    except Exception as e:
        print(f'Error Occured: {e}')
````

#### Master Data Cleaning Function

````python
# Master Preprocessing Function
def preprocess_apps_data():
    """
    Runs all preprocessing steps and returns:
    - Cleaned DataFrame
    """
    try:
        clean_rating_column()

        clean_reviews_column()

        convert_size_column()

        clean_installs_column()

        clean_price_column()

        split_genres()

        convert_last_updated()

        # Since we have new columns for processed 'Size', 'Last Updated' and 'Genres', we may drop the old columns
        apps.drop(['Size', 'Last Updated', 'Genres'], axis = 1, inplace = True)

    except Exception as e:
        print(f'Error Occured: {e}')

    print('Playstore Dataset Preprocessing Complete')
````

### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
correlation_matrix = apps[['Rating', 'Reviews', 'Installs', 'Price', 'Size_MB']].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of App Features')
plt.show()
````

1. Why did you pick the specific chart?
To show the correlation between the different numerical features of the apps in the dataset.
2. What is/are the insight(s) found from the chart?
- Reviews and Installs: There is a strong positive correlation (0.61) between the number of reviews and the number of installs. This suggests that apps with more installs tend to have more reviews, and vice versa. This is an expected relationship, as more users naturally lead to more feedback.
- Rating and other features: The rating of an app has a very weak positive correlation with the number of reviews (0.06) and installs (0.04). This indicates that a high rating alone doesn't necessarily guarantee a high number of installs or reviews.
- Price and other features: The price of an app has a very weak negative correlation with installs (-0.01) and reviews (-0.01). This suggests that paid apps don't necessarily have fewer installs or reviews than free apps, but the relationship is not strong.
Size and other features: The size of an app has a weak positive correlation with the number of reviews (0.08) and a very weak positive correlation with installs (0.03). This might suggest that larger apps, which may have more features, tend to get slightly more reviews, but the relationship with installs is negligible.
